```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Tugas Analisis 1

Hasil pada List biasa: Error (TypeError).
Penyebab: Operator * pada List Python berfungsi untuk mengulang/menggandakan isi list, bukan perkalian matematika. Mengalikan list dengan desimal (0.9) tidak diperbolehkan.
Perbedaan: NumPy Array didesain untuk operasi matematika, sehingga harga * 0.9 langsung mengalikan setiap angka di dalamnya.

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Tugas Analisis 2

Kolom bernilai kosong: menu (baris 4) dan terjual (baris 2).
Risiko tanpa pembersihan:
Analisis Salah: Rata-rata dan total penjualan menjadi tidak akurat.
Program Error: Fungsi perhitungan atau olah teks bisa gagal saat memproses nilai kosong (None/NaN).
Keputusan keliru: Informasi menu terlaris atau pendapatan kantin menjadi bias.

```python
# Membuat file data_kantin.csv sementara di Google Colab
df.to_csv('data_kantin.csv', index=False)

# Membaca data dari file CSV
df = pd.read_csv('data_kantin.csv')

print("--- 5 Baris Pertama ---")
print(df.head())

print("\n--- Info DataFrame ---")
print(df.info())

print("\n--- Statistik Ringkas ---")
print(df.describe())

print("\n--- Ukuran Dataset (Baris, Kolom) ---")
print(df.shape)
```

Tugas Analisis 3

Kolom dengan jumlah non-null lebih sedikit: Kolom yang nilai non-null count-nya kurang dari angka total baris pada range index (misalnya 95 non-null dari total 100 baris).

Artinya: Kolom tersebut memiliki data yang hilang atau kosong (missing values / NaN) pada baris-baris tertentu.

```python
print("--- Jumlah Data Kosong Sebelum ---")
print(df.isnull().sum())

# Penanganan missing values
df['terjual'] = df['terjual'].fillna(0)  # Isi kekosongan kolom terjual dengan 0
df = df.dropna(subset=['menu'])         # Hapus baris jika kolom menu kosong

print("\n--- Jumlah Data Kosong Sesudah ---")
print(df.isnull().sum())
```

Tugas Analisis 4

fillna(0) pada kolom `terjual`:
Alasan:`terjual` adalah data kuantitatif (angka). Mengisinya dengan `0` secara logis mengasumsikan bahwa item tersebut **tidak ada yang membeli** hari itu, sehingga angka $0$ tetap relevan untuk perhitungan statistik tanpa membuang informasi harga atau nama produknya.


dropna() pada kolom `menu`:
Alasan: `menu` adalah entitas utama (identitas produk). Tanpa nama menu, baris data tersebut tidak beridentitas—kita tidak tahu produk apa yang memiliki harga/penjualan tersebut, sehingga datanya menjadi **tidak bermakna/tidak berguna** untuk analisis bisnis dan lebih baik dihapus.

```python
print('Jumlah baris duplikat:', df.duplicated().sum())

# Menghapus duplikat
df = df.drop_duplicates()

# Memastikan tipe data integer
df['terjual'] = df['terjual'].astype(int)
df['harga'] = df['harga'].astype(int)

print("\n--- Tipe Data Kolom ---")
print(df.dtypes)
print("\n--- Data Setelah Duplikat Dihapus ---")
print(df)
```

Tugas Analisis 5

Jumlah baris: Berkurang dari 5 baris menjadi 3 baris (karena 1 baris menu None dihapus di langkah sebelumnya, dan 1 baris duplikat Es Teh dihapus).
Pentingnya tipe data (dtypes):
Akurasi Perhitungan: Mengubah teks menjadi angka (int) agar fungsi seperti sum() atau mean() bisa berjalan tanpa error.
Logika Data: Kolom terjual dan harga adalah jumlah barang dan nominal bulat, sehingga tidak relevan menggunakan koma/desimal (float).

```python
# 1. Filtering (terjual > 20)
laris = df[df['terjual'] > 20]
print("--- Menu Laris (>20) ---")
print(laris)
# 2. Sorting
urut = df.sort_values(by='terjual', ascending=True)
# 3. Kolom Turunan (total_pendapatan)
df['total_pendapatan'] = df['harga'] * df['terjual']
# 4. Agregasi Groupby
ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print("\n--- Ringkasan Total Pendapatan per Menu ---")
print(ringkasan)
print(urut)
```

Tugas Analisis 6

Menu pendapatan tertinggi: Nasi Goreng (dengan total pendapatan Rp276.000, hasil dari 12.000 × 23).

Manfaat untuk keputusan kantin:
Stok & Bahan Baku: Memastikan porsi dan bahan baku Nasi Goreng selalu tersedia lebih banyak agar tidak kehabisan.
Strategi Promosi/Bundling: Menjadikan Nasi Goreng sebagai menu utama untuk dipaketkan dengan menu yang kurang laku (misal: Paket Hemat Nasi Goreng + Es Teh).
Evaluasi Menu: Fokus mempertahankan kualitas menu penyumbang omzet terbesar bagi kantin.

```python
import pandas as pd

# 1. Menyiapkan DataFrame (hasil dari Latihan 1 - 5)
data_kantin = {
    'menu': ['Nasi Goreng', 'Mie Ayam', 'Es Teh'],
    'harga': [12000, 10000, 4000],
    'terjual': [23, 0, 40]
}
df = pd.DataFrame(data_kantin)

# 2. Filtering & Sorting
laris = df[df['terjual'] > 20]
urut = df.sort_values(by='terjual', ascending=False)

# 3. Membuat kolom turunan total_pendapatan
df['total_pendapatan'] = df['harga'] * df['terjual']

# 4. Agregasi dengan Groupby
ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print("--- Ringkasan Total Pendapatan per Menu ---")
print(ringkasan)

# 5. KODE TAMBAHAN: Mencari menu dengan total_pendapatan tertinggi secara otomatis
menu_tertinggi = ringkasan.idxmax()
pendapatan_tertinggi = ringkasan.max()

print("\n--- KESIMPULAN HASIL ANALISIS ---")
print(f"Menu dengan pendapatan tertinggi: {menu_tertinggi}")
print(f"Total pendapatan: Rp{pendapatan_tertinggi:,}")
```
